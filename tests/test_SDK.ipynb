```python
from modelgate import ModelGate

# 1. Initialize and Load (Handles downloading, temp files, and schema parsing automatically)
gate = ModelGate()
gate.load_model(
    model_path="https://huggingface.co/DmytroSerbeniuk/my-iris-model/resolve/main/model.joblib",
    schema="../default_schema.json"
)

# 2. Predict (Handles jsonschema validation, NumPy reshaping, and type normalization automatically)
prediction = gate.predict({
    "sepal_length": 5.1,
    "sepal_width": 3.5,
    "petal_length": 1.4,
    "petal_width": 0.2
})

# Output: "setosa"
print(prediction)
```
